Reads the CSV/PNG outputs written by `python scripts/run_pipeline.py`.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display, Markdown
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RES = ROOT / 'results'
display(Markdown((RES / 'summary.md').read_text()))
```

```python
m = pd.read_csv(RES / 'metrics_by_fold.csv')
m[m.target == 'rb'].pivot_table(index='horizon', columns='model', values='mae', aggfunc=['mean', 'std']).round(3)
```

```python
pd.read_csv(RES / 'rb_vs_best_baseline.csv').round(3)
```

```python
s = pd.read_csv(RES / 'sleep_policy.csv')
# stability of the conformal policy across folds
s[(s.horizon == 1) & s.policy.str.startswith('xgb_conformal')][['fold', 'policy', 'false_sleep_rate', 'capture', 'energy_saved_pct']].round(3)
```

```python
for f in ['horizon_curves_rb', 'calibration_rb', 'sleep_pareto_1h', 'example_forecast_1h', 'feature_importance_rb']:
    p = RES / 'figures' / f'{f}.png'
    if p.exists():
        display(Image(filename=str(p)))
```
